# Phân tích dữ liệu phân rã B23-B25
Dữ liệu mẫu do mô hình N(t) = N0*2**(-t/T) tạo ra, không phải dữ liệu đo. Thay N0, T, rồi so sánh ba giả thuyết. Chạy bằng Jupyter hoặc tải tệp lên Google Colab.
Đơn vị: t và T theo giây; N là số hạt.

```python
import math, csv
N0 = 800
T = 30
times = list(range(0, 121, 10))
counts = [N0 * 2**(-t/T) for t in times]
for t, n in zip(times, counts):
    print(f"t={t:3d} s, N={n:.1f}")
```

```python
# Chỉ ước lượng từ dữ liệu dương; không xử lí sai số đếm/nền trong ví dụ này.
estimates = [t * math.log(2) / math.log(N0/n) for t, n in zip(times, counts) if t > 0 and 0 < n < N0]
print("T ước lượng (s):", sum(estimates)/len(estimates))
with open("decay-model.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["t_seconds", "N_model"])
    writer.writerows(zip(times, counts))
```

```python
# Vẽ SVG bằng Python chuẩn, không cần cài thư viện đồ thị.
# Các đường thử T = 20, 30, 40 s; chấm đen là dữ liệu mô hình.
def xy(t, n):
    return 60 + t / 120 * 480, 330 - n / N0 * 280
svg = ['<svg xmlns="http://www.w3.org/2000/svg" width="650" height="400" viewBox="0 0 650 400">', '<rect width="650" height="400" fill="white"/>', '<path d="M60 40V330H560" fill="none" stroke="black"/>', '<text x="60" y="25">N (số hạt)</text>', '<text x="290" y="380">t (s)</text>']
for trial, color in [(20, "#be4242"), (30, "#137b65"), (40, "#286bb0")]:
    points = " ".join(f"{xy(t, N0*2**(-t/trial))[0]:.1f},{xy(t, N0*2**(-t/trial))[1]:.1f}" for t in range(121))
    svg.append(f'<polyline points="{points}" fill="none" stroke="{color}" stroke-width="2"/>')
    svg.append(f'<text x="430" y="{25+trial*2}" fill="{color}">T = {trial} s</text>')
for t, n in zip(times, counts):
    x, y = xy(t, n)
    svg.append(f'<circle cx="{x}" cy="{y}" r="3"/>')
for t in [0,30,60,90,120]:
    x, _ = xy(t, 0)
    svg.append(f'<text x="{x-8}" y="352">{t}</text>')
for n in [0,N0/2,N0]:
    _, y = xy(0, n)
    svg.append(f'<text x="12" y="{y+4}">{n:g}</text>')
svg.append('</svg>')
plot_svg = ''.join(svg)
with open('decay-comparison.svg', 'w', encoding='utf-8') as f:
    f.write(plot_svg)
try:
    from IPython.display import SVG, display
    display(SVG(plot_svg))
except ImportError:
    print('Đã tạo decay-comparison.svg; mở tệp để xem đồ thị.')
```
